# Rosen-Roback Spatial Equilibrium

> A modular framework for urban economics research on housing constraints and spatial misallocation


## What is this?

This library implements **spatial equilibrium models** to answer a simple question:

> **How much richer could we be with more liberal planning laws?**

Housing constraints in productive cities like London, New York, and San Francisco prevent workers from living where they're most productive. This **spatial misallocation** reduces GDP by billions.

We provide:

- 📐 **Model components** — Utility, housing supply, production, and equilibrium solvers
- 🔧 **High-level API** — Run counterfactual analyses with a few lines of code
- 📊 **Dataset abstractions** — Plug in UK, US, or synthetic data


## The Rosen-Roback Framework

The **Rosen-Roback model** is the workhorse of urban economics. It explains how workers and firms sort across cities based on:

| Factor        | Effect                                                  |
| ------------- | ------------------------------------------------------- |
| **Wages**     | Higher in productive cities (pulls workers in)          |
| **Rents**     | Higher where demand exceeds supply (pushes workers out) |
| **Amenities** | Weather, culture, safety (pulls workers to nice places) |

In **spatial equilibrium**, workers are indifferent between locations—any advantage in one dimension is offset by disadvantages in others.

### The Key Insight

When housing supply is **constrained** (low elasticity), productive cities can't grow. Workers who _would_ move to high-wage cities stay in lower-productivity areas because rents are too high. This misallocation costs GDP.

**Relaxing housing constraints** → More workers in productive cities → Higher GDP


## Quick Example


````python
from src.model import CounterfactualScenario, RosenRobackModel, ThreeCityDataset
from src.parameters import ModelParameters

# Create model with standard parameters
model = RosenRobackModel(params=ModelParameters.STANDARD)

# Use a simple three-city dataset
dataset = ThreeCityDataset(
    high_tfp=120,  # High productivity city
    medium_tfp=100,  # Medium productivity
    low_tfp=80,  # Low productivity
    high_elasticity=1.5,  # Constrained (like London/SF)
    low_elasticity=3.0,  # Elastic (like Houston)
)

# Solve baseline equilibrium
baseline = model.solve(dataset)
print(baseline.summary())
````

````python
# What if we liberalized housing in the high-productivity city?
scenario = CounterfactualScenario(
    name="Housing Reform",
    description="Double the housing supply elasticity in the productive city",
    elasticity_changes={"High Productivity": 3.0},  # From 1.5 to 3.0
)

baseline, counterfactual, gdp_impact = model.run_counterfactual(dataset, scenario)

print(f"GDP Impact: {gdp_impact['gdp_change_pct']:.2f}%")
print("\nPopulation shift to high-productivity city:")
print(f"  Before: {baseline.population[0]:,.0f}")
print(f"  After:  {counterfactual.population[0]:,.0f}")
````
